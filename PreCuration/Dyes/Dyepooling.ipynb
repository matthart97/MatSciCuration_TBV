# Data Pooling: Organic Dyes 

This shows the process of taking data from disprate resources and placing them into a single file. This will result in the dataset that we will be using for curation.

This entire process can also be done by running "DyePooling.py"

```python

import pandas as pd
import os


PATH = os.getcwd()
data_path = PATH +"/Dye_lake/"
```

```python
# load the data from Deep4Chem

dfc = pd.read_csv(data_path+'deep4chem.csv')
dfc
```

These properties are dependent on a sovent, so lets mark them as such

```python
#rename solvent dependet properties

exclude = ['SMILES','Tag']

dfc.rename(columns=lambda x: x + ('SolventDependent') if x not in exclude else x, inplace=True)
dfc
```

```python
# the tag is not needed 
dfc = dfc.drop(['Tag'], axis =1)
```

```python
dfc
```

Now we load the Ocelot database

```python
oclt = pd.read_csv(data_path+"ocelot_data.csv")
oclt
```

The columns are not named very well, so we will name them with less ambiguity

```python
oclt.columns = [
    'identifier', 'VerticalIonizationE', 'AdiabaticIonizationE', 'VerticalElectronAffinity', 'AdiabaticElectronAffinity', 'HLGap', 'lowest-lyingSinglet', 'lowest-lyingTriplet', 'HoleReorganizationE', 'CatIonRelax2', 'CationRelax1', 'ElectronReorganizationE', 'AnIonRelax1', 'AnionRelax2', 'lumo', 'homo', 'SMILES'
]
oclt
```

Finally, lets load the web-scraped data

```python
wdv = pd.read_csv(data_path+'WDVDataComplete_uncurated.csv')
wdv
```

like in  ocelot, some columns should be renamed to clear up ambiguity

```python

wdv.replace({1.0: 'yes', 0.0: 'unknown', -1.0: 'no'}, inplace=True)
```

```python
wdv
```

one last step: denote the origin of each data point before they are all merged into the same file

```python
wdv['origin'] = 'Webscraping'
oclt['origin'] = 'ocelot'
dfc['origin'] = 'Deep4Chem'
```

```python
# merge the dataframes into one and save
merged_df = dfc.merge(oclt, on=['SMILES','origin'], how='outer')
```

```python
merged_df.columns
```

```python
merged_df
```

```python
merged_df = merged_df.merge(wdv,on=['SMILES','origin'], how='outer')
```

```python
merged_df
```

```python
merged_df =merged_df.dropna(subset='SMILES')
merged_df
```

```python
# Export the summary table to make sure the data lake is ready for curation

DyeDescription = merged_df.describe()
```

```python
DyeDescription
```

```python
DyeDescription.to_csv('DyeDataLakeSummary.csv')
```
