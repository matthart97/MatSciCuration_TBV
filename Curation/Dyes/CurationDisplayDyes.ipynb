## Step four: Duplicate analyis  

There are two kinds of duplicate analysis we can do:

- feature-based
- string-based

We will start with string-based 

```python
# import needed packages and intiate the duplicate analyzer
from CurationScripts import DuplicateAnalysis 
analyzer = DuplicateAnalysis.DuplicateAnalyzer()
```

```python
# we will base the string-based analysis on SMILES strings
SMILES = NormalizedDf['NormalizedSmiles']
suspicousSMILES = analyzer.FindDuplcateSmilesStrings(SMILES,ReturnSuspiciousStrings=True)
```

```python
suspicious_smiles = analyzer.FindDuplcateSmilesStrings(SMILES, ReturnSuspiciousStrings=True)
```

```python
# inspect the the datapoints containing these strings

SmilesDuplicates = NormalizedDf['NormalizedSmiles'].isin(suspicious_smiles)
```

```python
# first, calculate features/descriptors for the whole dataset

from CurationScripts import GenericDescriptors
calculator = GenericDescriptors.GenericFeaturization()
```

```python
from scipy.spatial.distance import squareform, pdist

from tqdm import tqdm

from rdkit.Chem import AllChem
import pandas as pd 
from rdkit import Chem
from rdkit.Chem import AllChem
from typing import List
import numpy as np
```

```python
# use this code to find molecuels that are similar in feature space 


Fingerprint_matrix = F.filter(regex='fp_')
dist = pd.DataFrame(squareform(pdist(Fingerprint_matrix.iloc[:, 1:])))


# we have a potential duplicate when there is a very low distance between points
#change matrix to numpy for faster runtime
Dist = dist.to_numpy()
flaglist = []
#iterate over top triangle of distance matrix 
q = len(Dist)


for i in tqdm(range(1,q-1)):
  for j in range(i+1,q):
    z = Dist[i][j]
    if z == 0.0:
      flaglist.append((i,j))
    else:
      continue
     
```

```python
import pandas as pd
import numpy as np
from scipy.spatial.distance import pdist, squareform
from tqdm import tqdm

# Assume F is already defined and contains the necessary data
Fingerprint_matrix = F.filter(regex='fp_')
# Calculate pairwise distances directly as a numpy array
dist_array = squareform(pdist(Fingerprint_matrix))

flaglist = []
q = dist_array.shape[0]

# Iterate over top triangle of distance matrix
for i in tqdm(range(1, q-1)):
    for j in range(i+1, q):
        if dist_array[i, j] < 0.0:
            flaglist.append((i, j))

flaglist
```
